# Démonstration RAG avec l’API interne (sans HTTP)

```python
import asyncio
from src.services.rag import ask
from src.services.ingest import ingest

ingest(category="cs.CL", max_results=1)

async def demo():
    ans = await ask("Qu'est-ce qu'un modèle Transformer ?", top_k=3)
    print("Réponse:\n", ans["answer"])
    print("\nSources:")
    for s in ans["sources"]:
        print("-", s["title"], f"(score={s['score']:.2f})")

asyncio.run(demo())
```
